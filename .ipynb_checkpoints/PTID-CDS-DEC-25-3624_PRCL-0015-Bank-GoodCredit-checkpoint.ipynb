#### Predict cred score for current credit card customers. The cred score will denote a customer’s credit worthiness and help the bank in reducing credit default risk.
**Target variable → Bad_label**
- 0 – Customer has Good credit history
- 1 – Customer has Bad credit history (falls into 30 DPD + bucket)
  
**Benchmark model has gini as 37.9**

```python
# Step 1: Check for missing values and review data types
print(cd_df.isnull().sum())
print(cd_df.dtypes)

# Step 2: Convert date columns to datetime format
cd_df['dt_opened'] = pd.to_datetime(cd_df['dt_opened'], format='%d-%b-%y', errors='coerce')
cd_df['entry_time'] = pd.to_datetime(cd_df['entry_time'], format='%d-%b-%y', errors='coerce')

# Step 3: Handle missing dates
mean_opened_date = cd_df['dt_opened'].mean()
mean_entry_date = cd_df['entry_time'].mean()
cd_df['dt_opened'] = cd_df['dt_opened'].fillna(mean_opened_date)
cd_df['entry_time'] = cd_df['entry_time'].fillna(mean_entry_date)

# Step 4: Fill missing columns for features
cd_df['feature_3'] = pd.to_numeric(cd_df['feature_3'], errors='coerce')
cd_df['feature_3'] = cd_df['feature_3'].fillna(cd_df['feature_3'].mean())
cd_df['feature_4'] = cd_df['feature_4'].fillna(cd_df['feature_4'].mode()[0])

# Step 5: Optimize numeric column types
try:
    cd_df['feature_2'] = pd.to_numeric(cd_df['feature_2'], errors='coerce', downcast='integer')
except ValueError as e:
    print("Error converting feature_2 to numeric:", e)

# Step 6: Convert categorical columns to numeric using OneHotEncoder
categorical_cols = cd_df.select_dtypes(include='object').columns.tolist()
categorical_cols.remove('Bad_label')  # Assuming 'Bad_label' is the target variable

# One-hot encode categorical variables
# Use sparse_output instead of sparse
encoder = OneHotEncoder(sparse_output=True, drop='first')  # Change to sparse_output
encoded_features = encoder.fit_transform(cd_df[categorical_cols])
encoded_df = pd.DataFrame.sparse.from_spmatrix(encoded_features)
encoded_df.columns = encoder.get_feature_names_out(categorical_cols)

# Step 7: Combine encoded features with the rest of the DataFrame
cd_df = pd.concat([cd_df.drop(categorical_cols, axis=1), encoded_df], axis=1)

# Step 8: Prepare features and target variable
X = cd_df.drop('Bad_label', axis=1)
y = cd_df['Bad_label']

# Step 9: Convert DataFrame to sparse matrix
X_sparse = csr_matrix(X)

# Step 10: Apply PCA
pca = PCA(n_components=0.95)  # Explaining 95% of variance
X_pca = pca.fit_transform(X_sparse.toarray())  # Convert sparse to dense for PCA

print(f'Original number of features: {X.shape[1]}')
print(f'Reduced number of features after PCA: {X_pca.shape[1]}')

# Step 11: Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_pca, y, test_size=0.2, random_state=42)

# Step 12: Train a Random Forest model
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# Predictions and Evaluation
y_pred = model.predict(X_test)

# Print evaluation metrics
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))

# Optional: Visualize explained variance
plt.figure()
plt.plot(range(1, len(pca.explained_variance_ratio_) + 1), pca.explained_variance_ratio_, marker='o')
plt.title('Explained Variance by Principal Components')
plt.xlabel('Principal Components')
plt.ylabel('Variance Explained')
plt.show()
```
