# #### Predict cred score for current credit card customers. The cred score will denote a customer’s credit worthiness and help the bank in reducing credit default risk.
# **Target variable → Bad_label**
# - 0 – Customer has Good credit history
# - 1 – Customer has Bad credit history (falls into 30 DPD + bucket)
#
# **Benchmark model has gini as 37.9**

```python
# Step 1: Check for missing values and review data types
print(cd_df.isnull().sum())
print(cd_df.dtypes)
```

```python
cd_df['dt_opened'] = pd.to_datetime(cd_df['dt_opened'], format='%d-%b-%y', errors='coerce')
cd_df['entry_time'] = pd.to_datetime(cd_df['entry_time'], format='%d-%b-%y', errors='coerce')
```

```python
mean_opened_date = cd_df['dt_opened'].mean()
mean_entry_date = cd_df['entry_time'].mean()
```

```python
cd_df['dt_opened'] = cd_df['dt_opened'].fillna(mean_opened_date)
cd_df['entry_time'] = cd_df['entry_time'].fillna(mean_entry_date)
```

```python
cd_df['feature_3'] = pd.to_numeric(cd_df['feature_3'], errors='coerce')
cd_df['feature_3'] = cd_df['feature_3'].fillna(cd_df['feature_3'].mean())
cd_df['feature_4'] = cd_df['feature_4'].fillna(cd_df['feature_4'].mode()[0])
```

```python
try:
    cd_df['feature_2'] = pd.to_numeric(cd_df['feature_2'], errors='coerce', downcast='integer')
except ValueError as e:
    print("Error converting feature_2 to numeric:", e)
print("Initial data cleaning and type conversions re-executed for cd_df.")
```

```python
global irrelevant_cols
irrelevant_cols = [
    'customer_no',
    'feature_20',
    'feature_22',
    'feature_24',
    'feature_28',
    'feature_29',
    'feature_43',
    'feature_44',
    'feature_45',
    'feature_47',
    'feature_75',
    'feature_77'
]

all_object_cols = cd_df.select_dtypes(include='object').columns.tolist()
global categorical_cols_to_encode
categorical_cols_to_encode = [col for col in all_object_cols \
                              if col != 'Bad_label' and col not in irrelevant_cols]
print("irrelevant_cols and categorical_cols_to_encode lists re-defined.")
```

```python
print("Cardinality of categorical columns:")
for col in categorical_cols_to_encode:
    if col in cd_df.columns:
        cardinality = cd_df[col].nunique()
        print(f"Column '{col}': {cardinality} unique values")
    else:
        print(f"Warning: Column '{col}' not found in cd_df.")
```

```python
high_cardinality_threshold = 500  # Define a threshold for high cardinality

columns_to_remove_due_to_high_cardinality = []
for col in categorical_cols_to_encode:
    if col in cd_df.columns:
        cardinality = cd_df[col].nunique()
        if cardinality > high_cardinality_threshold:
            columns_to_remove_due_to_high_cardinality.append(col)

global categorical_cols_to_encode # Re-declare global to modify
categorical_cols_to_encode = [col for col in categorical_cols_to_encode if col not in columns_to_remove_due_to_high_cardinality]

print(f"Original number of categorical columns to encode: {len(cd_df.select_dtypes(include='object').columns.tolist()) - 1 - len(irrelevant_cols)}")
print(f"Removed high-cardinality columns: {columns_to_remove_due_to_high_cardinality}")
print(f"New number of categorical columns to encode: {len(categorical_cols_to_encode)}")
```

```python
encoder = OneHotEncoder(sparse_output=True, drop='first', handle_unknown='ignore')
encoded_features = encoder.fit_transform(cd_df[categorical_cols_to_encode])
print(f"Shape of new encoded_features: {encoded_features.shape}")
```

```python
# Step 8: Prepare features and target variable (dense part)
# Convert 'Bad_label' to numeric (0 or 1)
cd_df['Bad_label'] = pd.to_numeric(cd_df['Bad_label'], errors='coerce')

# Convert datetime columns to numerical representations (e.g., ordinal)
cd_df['dt_opened_numeric'] = cd_df['dt_opened'].apply(lambda x: x.toordinal() if pd.notnull(x) else 0)
cd_df['entry_time_numeric'] = cd_df['entry_time'].apply(lambda x: x.toordinal() if pd.notnull(x) else 0)

# Define columns to drop from cd_df to form the *dense* part of the feature matrix
# These include original datetime cols, irrelevant cols, and the categorical cols that were one-hot encoded
columns_to_drop_for_dense_part = ['dt_opened', 'entry_time'] + \
                                 [col for col in irrelevant_cols if col in cd_df.columns] + \
                                 [col for col in categorical_cols_to_encode if col in cd_df.columns]

# Create the dense part of the feature matrix DataFrame
X_dense_part_df = cd_df.drop(columns=columns_to_drop_for_dense_part, errors='ignore')

# Separate target variable 'y'
y = X_dense_part_df['Bad_label']
X_dense_part_df = X_dense_part_df.drop('Bad_label', axis=1)

# Ensure all columns in X_dense_part_df are numeric and handle NaNs for dense part
for col in X_dense_part_df.columns:
    if X_dense_part_df[col].dtype == 'object':
        X_dense_part_df[col] = pd.to_numeric(X_dense_part_df[col], errors='coerce')
X_dense_part_df = X_dense_part_df.fillna(0)

# The X_dense_part_df is now ready. The encoded_features (sparse) is also ready from 20-O68cwxEOh
```

```python
# Convert the dense part to a sparse matrix
X_dense_part_sparse = csr_matrix(X_dense_part_df.values)

# Combine with the one-hot encoded sparse features
# encoded_features was created after filtering high cardinality columns
X_sparse = hstack([X_dense_part_sparse, encoded_features])

# Print shape of the final sparse matrix
print(f"Shape of the final sparse feature matrix (X_sparse): {X_sparse.shape}")
```

```python
pca = PCA(n_components=0.95)  # Retain 95% of variance
X_pca = pca.fit_transform(X_sparse.toarray())  # Convert sparse to dense for PCA
print(f'Reduced number of features after PCA: {X_pca.shape[1]}')
```

```python
# Step 11: Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_pca, y, test_size=0.2, random_state=42)
```

```python
# Step 12: Train a Random Forest model
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)# Predictions and Evaluation
```

```python
y_pred = model.predict(X_test)
```

```python
# Print evaluation metrics
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```python
# Optional: Visualize explained variance
plt.figure()
plt.plot(range(1, len(pca.explained_variance_ratio_) + 1), pca.explained_variance_ratio_, marker='o')
plt.title('Explained Variance by Principal Components')
plt.xlabel('Principal Components')
plt.ylabel('Variance Explained')
plt.show()
```

```python
def preprocess_sample_data(sample_df_raw):
    # Ensure the sample_df_raw is a copy to avoid SettingWithCopyWarning
    sample_df = sample_df_raw.copy()

    # Convert 'Bad_label' to numeric (for consistency, even if not directly used in X)
    sample_df['Bad_label'] = pd.to_numeric(sample_df['Bad_label'], errors='coerce')

    # Convert datetime columns to numerical representations (e.g., ordinal)
    sample_df['dt_opened_numeric'] = sample_df['dt_opened'].apply(lambda x: x.toordinal() if pd.notnull(x) else 0)
    sample_df['entry_time_numeric'] = sample_df['entry_time'].apply(lambda x: x.toordinal() if pd.notnull(x) else 0)

    # Drop original datetime columns and other irrelevant columns (using global irrelevant_cols)
    # Also drop 'Bad_label' as it's the target
    columns_to_drop_for_dense_part_sample = ['dt_opened', 'entry_time', 'Bad_label'] + \
                                            [col for col in irrelevant_cols if col in sample_df.columns]

    X_sample_dense_part_df = sample_df.drop(columns=columns_to_drop_for_dense_part_sample, errors='ignore')

    # Ensure all columns in X_sample_dense_part_df are numeric and handle NaNs
    for col in X_sample_dense_part_df.columns:
        if X_sample_dense_part_df[col].dtype == 'object':
            X_sample_dense_part_df[col] = pd.to_numeric(X_sample_dense_part_df[col], errors='coerce')
    X_sample_dense_part_df = X_sample_dense_part_df.fillna(0)

    # Align columns of the sample dense part with the training dense part
    # This is crucial for consistent feature sets, especially before PCA
    # Use X_dense_part_df.columns from the global scope (from model training)
    if 'X_dense_part_df' in globals():
        original_dense_cols = X_dense_part_df.columns.tolist()
        # Add missing columns with 0 and remove extra columns from sample_dense_part
        for col in original_dense_cols:
            if col not in X_sample_dense_part_df.columns:
                X_sample_dense_part_df[col] = 0
        X_sample_dense_part_df = X_sample_dense_part_df[original_dense_cols]
    else:
        print("Warning: X_dense_part_df not found in global scope. Column alignment might be incorrect.")

    # One-hot encode categorical variables for the sample using the *fitted* encoder
    # Ensure all columns in categorical_cols_to_encode are present in sample_df before encoding
    missing_cols_in_sample = [col for col in categorical_cols_to_encode if col not in sample_df.columns]
    if missing_cols_in_sample:
        # Fill missing categorical columns with an empty string or a suitable placeholder
        # This ensures the encoder can transform them consistently if handle_unknown='ignore' is used
        for col in missing_cols_in_sample:
            sample_df[col] = '' # Adding them with a default empty string for consistency

    sample_encoded_features = encoder.transform(sample_df[categorical_cols_to_encode])

    # Convert the dense part of the sample to a sparse matrix
    X_sample_dense_part_sparse = csr_matrix(X_sample_dense_part_df.values)

    # Combine with the one-hot encoded sparse features
    X_sample_sparse = hstack([X_sample_dense_part_sparse, sample_encoded_features])

    # Apply PCA using the *fitted* pca object
    X_sample_pca = pca.transform(X_sample_sparse.toarray())

    return X_sample_pca

# Test the new function with the sample data created in the previous step
processed_sample_data = preprocess_sample_data(new_sample_df)
print(f"Shape of processed sample data after PCA: {processed_sample_data.shape}")
```

```python
def predict_credit_score(processed_data):
    # Make prediction using the trained model
    prediction = model.predict(processed_data)
    prediction_proba = model.predict_proba(processed_data)

    if prediction[0] == 0:
        cred_score_status = "Good credit history (Low risk of default)"
    else:
        cred_score_status = "Bad credit history (High risk of default)"

    print(f"The predicted credit score status for the example customer is: {cred_score_status}")
    print(f"Probability of Good Credit (0): {prediction_proba[0][0]:.4f}")
    print(f"Probability of Bad Credit (1): {prediction_proba[0][1]:.4f}")
    return cred_score_status, prediction_proba

# Call the functions to demonstrate the full pipeline
print("--- Predicting credit score for an example customer ---")
# 1. Create sample data
example_customer_df = create_sample_dataframe()

# 2. Preprocess sample data
processed_example_data = preprocess_sample_data(example_customer_df)

# 3. Predict credit score
status, probabilities = predict_credit_score(processed_example_data)
```
